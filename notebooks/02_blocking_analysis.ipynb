# 02 Blocking Strategy & Candidate Recall Analysis
Evaluate candidate generation recall and reduction ratio across all 11 blocking passes.

```python
import sys
import pandas as pd
sys.path.insert(0, '../src')
from business_entity_resolution.config import load_config
from business_entity_resolution.io import load_all_datasets, parse_ground_truth_matches
from business_entity_resolution.normalization import process_normalization
from business_entity_resolution.blocking import MultiStrategyBlocker
from business_entity_resolution.candidate_generation import evaluate_blocking_recall

config = load_config('../config.yaml')
datasets, df_gt = load_all_datasets(config, sample_size=5000)
gt_dict = parse_ground_truth_matches(df_gt)
df_s1 = process_normalization(datasets['s1'])
df_target = process_normalization(pd.concat([datasets['s2'], datasets['s3']], ignore_index=True))

blocker = MultiStrategyBlocker()
cands = blocker.generate_candidates(df_s1, df_target)
stats = evaluate_blocking_recall(cands, gt_dict, df_s1['entity_id'].tolist())
print(stats)
```
